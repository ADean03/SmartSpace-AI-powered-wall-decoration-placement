```python
import os
import yaml
from ultralytics import YOLO
import gradio as gr
from PIL import Image
```

```python
model_path = "/content/drive/My Drive/SmartSpace/testimages/best (2).pt"
model_path ="/best (2).pt"
model = YOLO(model_path)
```

```python
def run_yolo(image, conf):
    # Ensure confidence is float and within safe range
    try:
        conf = float(conf)
    except:
        conf = 0.5  # fallback default
    conf = max(0.0, min(1.0, conf))

    # Run inference
    results = model.predict(
        source=image,
        classes=[11],   # photo frames
        imgsz=416,
        conf=conf
    )

    # Extract plotted image
    result_img = Image.fromarray(results[0].plot())
    return result_img
```

```python
# Build Gradio UI
interface = gr.Interface(
    fn=run_yolo,
    inputs=[
        gr.Image(type="pil", label="Upload Image"),
        gr.Number(label="Confidence (0.0 to 1.0)", value=0.0007)
    ],
    outputs=gr.Image(label="Detection Result"), # allow_flagging=False
    title="SmartSpace Painting Detector/Placement predictor",
    description="Upload an image and set the detection confidence threshold. very small confidence values (0.00..) Recomended for predictions"
)

interface.launch(debug=True, share=True, inline=False)
```

```python
interface.close()
```
